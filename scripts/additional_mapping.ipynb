```python
import requests
import json

URL_SHIP_TYPE_DATA = "https://raw.githubusercontent.com/AzurLaneTools/AzurLaneData/refs/heads/main/KR/ShareCfg/ship_data_by_type.json"

def fetch_json_data(url):
    """Fetches JSON data from a given URL."""
    response = requests.get(url)
    response.raise_for_status()  # Raise an exception for bad status codes
    return response.json()

try:
    ship_type_data = fetch_json_data(URL_SHIP_TYPE_DATA)

    ship_type_conversion_dict = {}
    for key, value in ship_type_data.items():
        if key.isdigit(): # Ensure key is a digit before converting to int
            int_key = int(key)
            type_name = value.get("type_name")

            # Modify type_name based on key
            if key == "20":
                type_name = "미구-전열"
            elif key == "21":
                type_name = "미구-후열"
            elif key == "22":
                type_name = "범선-잠수"
            elif key == "23":
                type_name = "범선-전열"
            elif key == "24":
                type_name = "범선-후열"

            ship_type_conversion_dict[int_key] = {
                "ship_type": value.get("ship_type"),
                "type_name": type_name,
                "icon": f"https://raw.githubusercontent.com/Fernando2603/AzurLane/main/images/type/{key}.png"
            }

    print("Ship type conversion dictionary created and stored in 'ship_type_conversion_dict' variable.")
    print(list(ship_type_conversion_dict.items())[:5])

    with open('ship_type_mapping.json', 'w', encoding='utf-8') as f:
        json.dump(ship_type_conversion_dict, f, ensure_ascii=False, indent=4)
    print("ship_type_conversion_dict saved to 'ship_type_mapping.json'")

except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
except ValueError as e:
    print(f"Error parsing JSON: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```
